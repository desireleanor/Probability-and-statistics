```python
import pandas as pd
student_data = {
    "Name": ["John", "Mary", "Steve", "Sarah", "David", "Anna"],
    "Age": [23, 13, 56, 34, 12, 41],
    "Maths": [90, 80, 70, 60, 50, 40],
    "English": [85, 75, 65, 55, 45, 35]
}
students = pd.DataFrame(student_data)
students
```

1. Find the average of english.

```python
students["English"].mean()
```

The average of English is 60.0

2. Find the average of mathematics.

```python
students["Maths"].mean()
```

The average of Maths is 65.0

3. Find the median age.

```python
students["Age"].median()
```

The median age is 28.5

4. Find the median mark of Maths

```python
students["Maths"].median()
```

The median mark of Maths is 65.0

5. Find the standard deviation for english.

```python
students["English"].std()
```

The standard deviation of English is 18.708

6. What is the range of the age values in our data?

```python
range = students["Age"].max() - students["Age"].min()
range
```

The range of the age values is 44

7. Which student scored the highest mark in English?

```python
students.loc[students["English"].idxmax(), "Name"]
```

The student with the highest marks is John
