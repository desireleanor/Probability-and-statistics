## Probability Theory
### Probability:
This is the likelihood of an event occurring.
 

## Fundamental Concepts of Probability
### 1. Random Experiment
Is a process whose outcome can not be predicted with certainty before it is performed. Examples include:
- Tossing a coin.
- Rolling a die.
- Selecting a random user from a database.
### 2. Sample Space
Is a set of all possible outcomes from a random experiment.
Examples include:
- Tossing a coin: `S = {H, T}`
- Tossing two coins: `S = {HH, HT, TH, TT}`
- Roll a die: `S = {1, 2, 3, 4, 5, 6}` 
- Rolling two dice: `S = {(1,1), (1,2), (1,3), (1,4), (1,5), (1,6), (2,1),...,(6,6)}`

Tossing a single coin

```python
coin = ['H', 'T']

print(coin)
print(len(coin))
```

Tossing two coins

```python
from itertools import product

coin = ['H', 'T']
two_coins = list(product(coin,repeat=2))

print(two_coins)
print(len(two_coins))
```

Rolling a die

```python
one_die = [1, 2, 3, 4, 5, 6]

print(one_die)
print(len(one_die))
```

Rolling two dice

```python
one_die = [1, 2, 3, 4, 5, 6]
two_dice = list(product(one_die, repeat=2))

print(two_dice)
print(len(two_dice))
```

### 3. An Event
Is a specific outcome or a combination of several outcomes.
An event is a subset of the sample space.

Examples: 
- An event of getting a head when you toss a coin.
- An event of getting two tails when you toss two coins.
- An event of getting an even number when you roll a die.
- An event of getting two even numbers when you roll two dice.



An event of getting a head when you toss a coin

```python
one_coin = ['H', 'T']

event_head = [a for a in one_coin if a == 'H']

print(event_head)
print(len(event_head))
```

An event of getting an even number from sample space of rolling one dice

```python
one_dice = [1, 2, 3, 4, 5, 6]

event_even = [x for x in one_dice if x % 2 == 0]

print(event_even)
print(len(event_even))
```

An event for getting an odd number from sample space of rolling one die

```python
event_odd = [y for y in one_dice if y % 2 != 0]

print(event_odd)
print(len(event_odd))
```

An event for an outcome greater than 3 from a sample space of rolling one die

```python
event_three = [x for x in one_dice if x > 3]

print(event_three)
print(len(event_three))
```

An event of getting two heads when two coins are tossed

```python
coin = ['H', 'T']
two_coins = list(product(coin,  repeat=2))

event_two_heads = [(a, b) for a, b in two_coins if (a, b) == ('H','H')]

print(event_two_heads)
print(len(event_two_heads))
```

An event of getting a sum of 6 when two dice are rolled

```python
dice = [1, 2, 3, 4, 5, 6]
two_dice = list(product(dice, repeat=2))

event_sum_six = [(x, y) for x, y in two_dice if x + y == 6]

print(event_sum_six)
print(len(event_sum_six))
```

#### Probabilities of events.
The probability of an event A, written as `P(A)`, is a number between 0 and 1 that measures the likelihood that A occurs.

The probability of an event A, `P(A)` is given by:
$$ 
P(A) = \frac {\text{Prefered}\ \text{outcome}(\text{event})}{\text{All}\ \text{possible}\ \text{outcomes}(\text{sample}\ \text{space})}
$$

Example:
 - The probability of getting a tail when a coin is tossed.


```python
coin = ['H', 'T']
event_tail = [t for t in coin if t == 'T']

prefered_outcome = len(event_tail)
sample_space = len(coin)

P_tail = prefered_outcome / sample_space
print(f"The probability for getting a tail is: {P_tail}")
```

 - The probability of getting a sum of 6 when two dice are thrown.

```python
die = [1, 2, 3, 4, 5, 6]
two_dice = list(product(dice, repeat=2))

event_sum_eight = [(x, y) for x, y in two_dice if x + y == 8]

prefered_outcome = len(event_sum_eight)
sample_space = len(two_dice)

P_eight = prefered_outcome / sample_space
print(f"The probability is: {P_eight:.3f}")
```

#### Set operations with Events
Events can be combined using set operations.\
For given sets `A` and ` B`, we can have operations such as: 
$$ \text{Intersection:}\  \text{A} \cap \text{B} $$
$$ \text{Union:}\  \text{A} \cup \text{B} $$
$$ \text{Complement:}  A^c $$
$$ \text{Null:}\  \varnothing$$
$$ \text{Subset: }\  \text{A} \subseteq \text{B} $$



Note: For Intersection between two sets, we use `&` and for Union, we use `|`.

```python
S = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {1, 3, 5}
C = {4, 5, 6}

print(f"The Intersection of A & S is: {A & S}")
print(len(A & S))

print(f"\nThe Union of A & B is: {A | B}")
print(len(A | B))

print(f"\nThe compliment of A is: {S - A}")
print(len(S - A))

print(f"\nA and B are disjoint sets: {len(A & B) == 0}")

print(f"\nIs B a subset of S?: {B.issubset(S)}")
```

##### Application to a Real dataset

Example 1: Using the housing dataset, Find the probability of a randomly selected house having more than 3 bedrooms and more than 2 bathrooms.

```python
housing = pd.read_csv('Housing.csv')

event_beds_and_baths = (housing['bedrooms'] > 3) & (housing['bathrooms'] > 2)

prefered_outcome = event_beds_and_baths.sum()
sample_space = housing.shape[0]

P_event = prefered_outcome / sample_space
print(f"The probability is: {P_event:.4f}")
```

Example 2: Using the housing dataset, Find the probability of a randomly selected house having more than 3 bedrooms or more than 2 bathrooms. 

```python
event_beds_or_baths = (housing['bedrooms'] > 3) | (housing['bathrooms'] > 2)

prefered_outcome = event_beds_or_baths.sum()
sample_space = housing.shape[0]

P_event = prefered_outcome / sample_space
print(f"The probability is: {P_event:.4f}")
```
