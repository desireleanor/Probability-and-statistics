### 1. How do sales change over time?

```python
# Creating a new column for the year of the order date to analyze sales over time
company_data['Order Date'] = pd.to_datetime(company_data['Order Date'])
company_data['Year'] = company_data['Order Date'].dt.year
company_data
```

```python
# Calculating total sales for each year
yearly_sales = company_data.groupby('Year')['Sales'].sum()
yearly_sales
```

```python
# Creating a line graph to visualize the total sales over the years
yearly_sales.plot(figsize=(10, 6), title='Total Sales Over the Years', xlabel='Year', ylabel='Total Sales')
```

    The line graph shows the trend in total sales from 2016 to 2019. Sales increased from approximately 11.8 million in 2016 to a peak of about 19 million in 2017, indicating strong growth during this period. However, sales began to decline in 2018 to approximately 17.2 million, followed by a much sharper decrease in 2019 to around 7.3 million, the lowest value recorded. Overall, the graph shows that sales performance was not consistent throughout the period, highlighting the need to investigate the factors that may have contributed to the decline.

### 2. What is the relationship between price and sales?

```python
fig = px.scatter(
    company_data,
    x='Price',
    y='Sales',
    hover_name='Product Sub Category',
    title='Relationship Between Price and Sales'
)

fig.show()
```

         The scatter plot shows a generally positive relationship between price and sales, as higher-priced products tend to have higher sales values.
    However, the points are widely scattered, indicating that price alone does not fully determine sales.
    Most observations are concentrated at lower prices and lower sales, while fewer products occur at higher price levels.
    There are also some unusually high sales values, suggesting the presence of potential outliers.
    Overall, the relationship appears positive but relatively variable, meaning other factors may also influence sales.

### 3. Which catagories have the highest sales?

```python
sales_by_category = (

     company_data.groupby('Product Sub Category')['Sales']

     .sum()

     .sort_values(ascending=False)

)

plt.figure(figsize=(12, 6))

sales_by_category.plot(kind='bar', color='steelblue', edgecolor='black')

plt.title('Total Sales by Product Sub Category')

plt.xlabel('Product Sub Category')

plt.ylabel('Total Sales')

plt.xticks(rotation=45, ha='right')

plt.tight_layout()

plt.show()
```

    The distribution of sales across the product sub-categories suggests that customer demand is not evenly spread across all products.The graph shows that sales are concentrated in a few product sub-categories, while several others contribute relatively little to total sales.This is because the stronger performance of categories such as Camcorders, Projectors & Screens, and Laptops indicates that these products may have greater demand or contribute more significantly to the business's overall sales. In contrast, the much lower sales from categories such as VCD & DVD and Bluetooth Headphones suggest relatively lower demand within the dataset.

### 4. How are the product prices distributed?

```python
mean_price = company_data['Price'].mean() 
median_price = company_data['Price'].median()
mode_price = company_data['Price'].mode()[0] 
print(f"Mean Price:   {mean_price}")
print(f"Median Price: {median_price}") 
print(f"Mode Price:   {mode_price}")   

company_data['Price'].plot(kind='hist', bins=20, figsize=(12, 5), title='Distribution of Product Prices', color='maroon', edgecolor='black', density=True)
company_data['Price'].plot.kde(color='yellow', label='KDE')

plt.axvline(mean_price, color='orange', linestyle='--', linewidth='1.5', label=f'Mean Price: {mean_price:.2f}',) 
plt.axvline(median_price, color='green', linestyle='--', linewidth='1.5', label=f'Median Price: {median_price:.2f}',)
plt.axvline(mode_price, color='blue', linestyle='--', linewidth='1.5', label=f'Mode Price: {mode_price:.2f}') 

plt.legend()
plt.show()
```

    The distribution of product prices is a predominantly unimodal, positively skewed (right-skewed) distribution.It is unimodal because it has one dominant peak  and the distribution has a long tail to the right thus being positively skewed.The mean price of 296.51 being higher than the median price of 205.00, indicating that the higher-priced products influence the average price. The distribution suggests that relatively low-priced products are more common in the dataset, while expensive products occur less frequently.

### 5. What is the distribution of sales values, and are there any unusually high or low sales outliers?

```python
plt.figure(figsize=(12, 6))

plt.boxplot(
    company_data['Sales'],
    orientation='horizontal',
    showfliers=True
)

plt.title('Distribution of Sales Values and Outliers')
plt.xlabel('Sales')

plt.tight_layout()
plt.show()
```

    - The boxplot shows that the Sales values are concentrated mainly at the lower end, with the median at approximately 2,301.18. The distribution has a long spread toward higher sales values, with numerous observations appearing beyond the upper whisker.

    - These observations represent potential high-value outliers. The maximum Sales value is 78,312, indicating that some sales values are considerably higher than the majority of the dataset. There are no potential low outliers, as the lower whisker does not show observations falling unusually far below the main distribution.

### 6. What proportions of Products belong to each category?

```python
sales_by_subcat = company_data.groupby(
    'Product Sub Category'
)['Sales'].sum().sort_values(ascending=False)

# Calculate percentage of total sales
percentages = (sales_by_subcat / sales_by_subcat.sum()) * 100

plt.figure(figsize=(12, 8))

plt.pie(
    sales_by_subcat.values,
    labels=None,
    startangle=90,
    shadow=True
)

plt.title('Proportion of Total Sales by Product Sub Category')

# Create legend labels containing category + percentage
legend_labels = [
    f"{category} — {percentage:.1f}%"
    for category, percentage in zip(
        sales_by_subcat.index,
        percentages
    )
]

plt.legend(
    legend_labels,
    title='Product Sub Category',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.tight_layout()
plt.show()
```

    - The pie chart shows how total sales are distributed across the different product sub-categories. Camcorders contribute the largest proportion of total sales at 15.3%, followed by Projectors & Screens at 12.7% and Laptops at 11.6%. Digital SLR Cameras also make a significant contribution of 10.3%. Together, these categories account for a substantial proportion of the total sales.

    - On the other hand, several categories contribute relatively small proportions of total sales. For example, VCRs & DVD contributes only 0.4%, Bluetooth Headphones 0.5%, and Home & Office Phones 0.6%. This indicates that sales are distributed across many product categories, although some categories contribute considerably more to total sales than others.

    - This comparison is useful because it shows which product sub-categories contribute most to the company's overall sales. It can help the business identify its major sources of sales and compare them with categories that contribute relatively little.
