## PANDAS
### Pandas Data Structures

### 1. pandas series is a one dimensional array

```python
import pandas as pd                    

ageOfStudents = pd.Series(
    [23, 13, 56, 34, 12, 41],
    index = ['John', 'Mary', 'Steve', 'Sarah', 'David', 'Anna'],
    name = 'Age of Students'                    #we can change the index
    
)
ageOfStudents 
```

### A dataframe is a 2-dimensional array.
This is to mean that it contains rows and columns

```python
student_data = {
    "Name": ["John", "Mary", "Steve", "Sarah", "David", "Anna"],
    "Age": [23, 13, 56, 34, 12, 41],
    "Maths": [90, 80, 70, 60, 50, 40],
    "English": [85, 75, 65, 55, 45, 35]
}
students = pd.DataFrame(student_data)
students
```

```python
students["English"]
```

1. what is the modal age?

```python
students["Age"].mode()[0] #gives value at index 0 of the mode series
```

From the results above, the modal age of students is 12

```python
students["Maths"].var()
```

The range is 350

```python
students["Maths"].std()
```

The standard deviation is 18.71

```python
range = students["Maths"].max() - students["Maths"].min()
range
```

The range is 50
