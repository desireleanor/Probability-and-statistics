```python
import pandas as pd
```

```python
titanic = pd.read_csv("titanic.csv")
```

```python
titanic.head(10)
```

```python
data = titanic[["survived", "pclass","sex","age","alone","embark_town"]].copy()
data
```

Find the number of passengers that:
    - survived
    -are male
    -travelled alone
    -travelled infirst class

```python
survived= data["survived"] == 1
print(f"survived number is {survived.sum()}")
```

```python
male = data["sex"] == "male"
male = male.sum()
print(f"Male number is {male}")
```

Find the probability of that:
    -survived

```python
preferred = survived.sum()
sample_space= data.shape[0]
prob = preferred/sample_space
print(f"The probability is {prob:.4f}")
```

Alternatively

```python
alone = data["alone"] == True
prob(alone)
```

### Joint Events

How many passengers embarked on southhampton and travelled alone

```python
embark_southhampton = data["embark_town"]== "Southampton"
alone= data["alone"] == True
joint_event = embark_southhampton & alone
joint_event.sum()
```

What is the probability for the above

```python
prob = prob(joint_event) 
print(f"The probability is {prob:.4f}")
```

The Contingency table(Frequency Table)  #two categorical values

```python
joint = pd.crosstab(data["sex"],data["survived"], normalize=True)
joint.columns = ["did not survive", "survived"]
joint
```

## Mutually Exclusive

```python
male = data["sex"] == "male"
female = data["sex"] == "female"
union = male|female
```

Calculate the individual probabilities of events and the union
Make a conclusion

```python
P_male = prob(male)
P_female = prob(female)
P_union = prob(union)

sum_prob = P_male + P_female

print(sum_prob)
print(P_union)  #since both sum and union are 1, the events are mutually exclusive
```

Using titanic, implement a scenario to check for independent events

```python
male = data["sex"] == "male"
survived = data["survived"] == 1
male_and_survived = male & survived
```

```python
P_male = prob(male)
P_survived = prob(survived)
P_male_and_survived = prob(male_and_survived)
```

```python
product = P_male * P_survived
print(product)
print(P_male_and_survived)  #since both product and P_male_and_survived are not equal, the events are not independent
```

From titanic, find the probability that the passenger survived given that is male(Conditional)

```python
prob_survived_given_that_is_male = P_male_and_survived / P_male
print(f"The probability is {prob_survived_given_that_is_male:.4f}")
```
