```python

def find_outliers(group):
    Q1 = group["body_mass_g"].quantile(0.25)
    Q3 = group["body_mass_g"].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = group[(group["body_mass_g"] < lower) |
                     (group["body_mass_g"] > upper)]
    
    return pd.Series({
        "IQR": IQR,
        "Number of Outliers": len(outliers)
    })

result = data3.groupby("species").apply(find_outliers)

print(result)
```
