1. Find the mean math score.

```python
student_data["math score"].mean()
```

2. What is the median reading score.

```python
student_data["reading score"].median()
```

3. What is the modal writing score.

```python
student_data["writing score"].mode()[0]
```

4. Find the average math score for females.

```python
student_data[student_data['gender']=='female']["math score"].mean()
```

5. Construct a new column called total(summation of math, reading and writing scores) and find its average.

```python
student_data["Total"] = student_data["math score"] + student_data["reading score"] + student_data["writing score"]
student_data["Total"].mean()
```

### Measures of position
## Quartiles

```python
student_data["math score"].quantile(0.25)  #lower quartile == 0.25
```

```python
student_data["math score"].quantile(0.5)  #middle quartile(median) == 0.5
```

```python
student_data["math score"].quantile(0.75)   #upper quartile == 0.75
```

```python
#interquartile range
student_data["math score"].quantile(0.75) - student_data["math score"].quantile(0.25)  
```
