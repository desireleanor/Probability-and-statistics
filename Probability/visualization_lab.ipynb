## Visualization of Data

## Line graphs
These are used to show variations/trends over time. For example
  - Number of bicycle hires over a period of time.
  - Growth of a population over time.

Let us import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Let us create a own DataFrame.

```python
data = {
    'Year' : [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020],
    'No of Bicycle Hires' : [2180813, 7142449, 9519283,8045459, 10023897, 9871839,10303637, 10446044,10567540,10424955,10434167]
}

bicycle_hires = pd.DataFrame(data)
bicycle_hires
```

We can then plot a line graph showing bicycle hires over a certain period of time

### Plotting the Line graph using the Pandas Version


  - The `plot`method includes all numeric `Dataframe` columns by default.
  - To plot a specific column of values, use the `y` parameter. As shown below we are plotting No of Bicycle hires

```python
bicycle_hires.plot(x= 'Year', y = 'No of Bicycle Hires',
                    figsize=(10,5),linewidth=2,linestyle="dotted", 
                    title="BICYCLE HIRES OVER TEN YEARS")

plt.tight_layout()
plt.show()
```

## The Histogram
 A Histogram is used to show the distribution of values. It is a graph showing the number of appearances of a value within a certain range. 
- The distribution tells you how frequently values fall into different ranges.
- The `bins=` parameter controls how many bars the range is divided into.

Let us load the housing dataset.

```python
housing_data = pd.read_csv('Housing.csv')
housing_data
```

We can now plot a Histogram showing the price distribution of the houses.
On the histogram, we also visualize the Mean, Median, Mode of the Distribution as well as the KDE curve.

```python
mean_value = housing_data["price"].mean()
median_value = housing_data["price"].median()
modal_value = housing_data["price"].mode()[0]

housing_data.plot.hist(y="price", figsize=(10,5), bins=20, color="steelblue",
                        edgecolor="black", title="HISTOGRAM",density=True )

housing_data["price"].plot.kde( color="red")

plt.axvline(mean_value, color="red", linestyle="--",label= f"Mean Value: {mean_value:.1f}")
plt.axvline(median_value,color="black", linestyle="--",label= f"Median Value: {median_value:.1f}" )
plt.axvline(modal_value, color="orange", linestyle="--", label = f"Modal Value: {modal_value:.1f}" )


plt.legend()
plt.show()
```

From the histogram above, we can determine if the data is positively skewed, negatively skewed or symmetric.
- Positive Skewness (Right-Skewed): The tail extends to the right (towards higher values). For housing prices, most homes are clustered in the lower to moderate price range, while a few very expensive homes pull the tail to the right. Typically, $\text{mean} > \text{median} > \text{mode}$.
- Negative Skewness (Left-Skewed): The tail extends to the left (towards lower values). For housing prices, most homes are clustered at higher values, while a few unusually cheap homes pull the tail to the left. Typically, $\text{mean} < \text{median} < \text{mode}$.
- Symmetric: The distribution is balanced on both sides of the central peak, where $\text{mean} \approx \text{median} \approx \text{mode}$ and both tails extend equally.


Alternatively, you can supply the `plot` method with the `kind` parameter to which you specify the type of chart you want to visualize. Here we want to visualize a Histogram.

```python
mean_value = housing_data["price"].mean()
median_value = housing_data["price"].median()
modal_value = housing_data["price"].mode()[0]

# This is the Alternative line  
housing_data.plot(kind='hist', y='price', bins=20, 
                    edgecolor='black', figsize=(10,5),
                    color="steelblue", title="HISTOGRAM",density=True)


housing_data["price"].plot.kde( color="red")

plt.axvline(mean_value, color="red", linestyle="--",label= f"Mean Value: {mean_value:.1f}")
plt.axvline(median_value,color="black", linestyle="--",label= f"Median Value: {median_value:.1f}" )
plt.axvline(modal_value, color="orange", linestyle="--", label = f"Modal Value: {modal_value:.1f}" )

plt.legend()
plt.show()
```

## Scatter Plot
A scatter plot is a graph where each point represents an observation's values on two variables(x, y). It visually reveals the relationship between two quantitave variables.
- Each row becomes a point, with one column on the x-axis and another on the y-axis.
- Scatter plots can show outliers and clusters in a dataset.



```python

housing_data.plot.scatter(x='area', y='price',
                        figsize=(12,6), color='red', marker='*',
                        title="A Scatter Plot showing the relationship between area and price of the Houses",
                        fontsize=20,xlabel="Area", ylabel="Prices" )

plt.show()
```

## Box Plot
   A box plot graphically shows groups of numerical data through their quartiles.The box extends from the Q1 to Q3 quartile values of the data, with the line at the median(Q2).
   - The median(Q2) is the middle value of the dataset and is represented by a line inside the box.
   - Q1 is the first quartile, a value below which 25% of the data falls.
   - Q3 is the third quartile, a value below which 75% of the data falls.
   - The Interquartile Range(IQR) is the distance between Q1 and Q3 representing the middle 50% of the data.
   - The whiskers(lines) extend from the edges of the box to show the range of most data values.
   - Outliers are plotted as separate dots and fall far from the rest of the data.

   Box plots are used to:
   - Understand the distribution of the data.
   - Identify the median and show data is spread.
   - Detect outliers.
   - Compare distribution across variables.
   - Check whether data is skewed or not.


```python
plt.figure(figsize=(10, 5))
sns.boxplot(housing_data['bathrooms'],showfliers=True)

plt.title('A Box Plot showing the distribution of Bathrooms')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(housing_data[['bathrooms', 'bedrooms']],showfliers=True)

plt.title('A Box Plot showing the Comparison of the distribution of Bathrooms and Bedrooms')
plt.tight_layout()
plt.show()
```

Alternatively, we can as well use Pandas approach as shown below. 

```python
housing_data.plot.box(y='bathrooms')

plt.show()
```

## Bar Charts
 A bar chart uses bars to represent occurrences/counts of values/categories.
 - Bars can be displayed vertically or horizontally.
 - Each bar represents a different category.
 - The height of the bar is proportional to the value it represents.

Let us create a dataframe called revenue that has the revenue for three different cities collected on three specific dates.

```python
revenue = pd.DataFrame({
    'Date':['1/4/2026', '2/4/2026', '3/4/20267'],
    'Kampala':[10000, 20000, 30000],
    'Wakiso':[25000, 35000, 15000],
    'Mukono':[40000, 20000, 15000]

})
revenue = revenue.set_index('Date')
revenue
```

```python
revenue.plot.bar (stacked=True)

plt.tight_layout()
```

## Pie Charts
A pie chart is a circle that is divided into sectors where each sector represents a category's proportion of the whole dataset.
- Each sector shows a category's proportion or percentage.
- The size of each sector depends on its value.
- All sectors together add up to 100%.

Let us draw a pie chart showing the area covered by the first five houses in the housing dataset. 

```python
housing_data = housing_data.head()
housing_data['area'].plot(kind='pie', legend=True, autopct='%1.1f%%')

plt.tight_layout()
plt.show()
```
