## Line Graph


Import the necessary Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Year": [2020, 2022, 2024, 2026, 2028, 2030],
    "Cars": [243596, 106888, 2990476, 50000, 3050007, 4099802]
}

cars_sold = pd.DataFrame(data)
cars_sold
```

```python
cars_sold.plot(x="Year",y="Cars")
```

Specify the size

```python
cars_sold.plot(x="Year",y="Cars", figsize=(10,5),linewidth=2,linestyle="dotted", title="Car Sales over ten years", ylabel="CARS")
plt.show()
```

### Histogram
This is used to show how data is distributed


```python
housing_data = pd.read_csv("Housing.csv")
housing_data.head()
```

Identify columns that are numerical and categorical

```python
mean_value = housing_data["price"].mean()
print(mean_value)

median_value = housing_data["price"].median()
print(median_value)

mode_value = housing_data["price"].mode()[0]
print(mode_value)

housing_data.plot.hist(y="price",figsize=(10,5),bins=20, color="steelblue", edgecolor="black", title="HISTOGRAM", density=True)     #specify the cloumns needed ,bins is the vertical scale
housing_data["price"].plot.kde(color="black")   #brings a curve over the histogram with association to density=True
                                                #positively skewed or right skewed(look at the mean and median: mean is larger than median )
                                                #kde
plt.axvline(mean_value, color="red", linestyle="--", label= f"Mean Value: {mean_value:.1f}")
plt.axvline(median_value, color="green", linestyle="--", label= f"Median Value: {median_value:.1f}")
plt.axvline(mode_value, color="pink", linestyle="--", label= f"Modal Value: {mode_value:.1f}")

plt.legend()
plt.show()
```

### Scatter graph

```python
housing_data.plot.scatter(x="bedrooms", y="bathrooms",marker="*")
plt.show()
```
